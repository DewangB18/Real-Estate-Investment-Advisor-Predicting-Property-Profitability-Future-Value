```python
# ==================== EDA ANSWERS: QUESTIONS 1–20 ====================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Automatically load the dataset if not already in memory
if "df_clean" not in globals() and "df" not in globals():
    try:
        df = pd.read_csv("/content/india_housing_prices.csv")
        print("Loaded /content/india_housing_prices.csv into df.")
    except Exception as e:
        pass

# Prefer the cleaned dataset created in earlier notebook cells.
if "df_clean" in globals() and isinstance(df_clean, pd.DataFrame):
    eda = df_clean.copy()
elif "df" in globals() and isinstance(df, pd.DataFrame):
    eda = df.copy()
else:
    raise NameError("Load the project's india_housing_prices.csv into df (or create df_clean) before running this EDA section.")

# Normalize whitespace in text columns without changing original dataframe.
for c in eda.select_dtypes(include=["object"]).columns:
    eda[c] = eda[c].map(lambda x: x.strip() if isinstance(x, str) else x)

def has(*cols):
    missing = [c for c in cols if c not in eda.columns]
    if missing:
        print("Cannot calculate this item; missing column(s):", missing)
        return False
    return True

def numeric(col):
    return pd.to_numeric(eda[col], errors="coerce")

def section(q, title):
    print("\n" + "="*90)
    print(f"QUESTION {q}: {title}")
    print("="*90)

# Derive rupees/sq ft explicitly. Retain stored Price_per_SqFt for comparison.
if has("Price_in_Lakhs", "Size_in_SqFt"):
    eda["Calculated_Rupees_Per_SqFt"] = (
        numeric("Price_in_Lakhs") * 100000 /
        numeric("Size_in_SqFt").where(numeric("Size_in_SqFt") > 0)
    )
else:
    eda["Calculated_Rupees_Per_SqFt"] = np.nan

# Q1. Distribution of property prices
section(1, "What is the distribution of property prices?")
if has("Price_in_Lakhs"):
    s = numeric("Price_in_Lakhs").dropna()
    print(f"Valid records: {len(s):,}; missing/non-numeric: {len(eda)-len(s):,}")
    print(f"Mean: {s.mean():,.2f} lakhs | Median: {s.median():,.2f} lakhs")
    print(f"Std dev: {s.std():,.2f} lakhs | Min: {s.min():,.2f} | Q1: {s.quantile(.25):,.2f} | Q3: {s.quantile(.75):,.2f} | Max: {s.max():,.2f}")
    print("Interpretation: mean substantially above median suggests a right-skewed distribution; confirm from the histogram.")
    sns.histplot(s, bins=30, kde=True); plt.title("Q1. Property price distribution"); plt.xlabel("Price (lakhs)"); plt.show()

# Q2. Distribution of property sizes
section(2, "What is the distribution of property sizes?")
if has("Size_in_SqFt"):
    s = numeric("Size_in_SqFt").dropna()
    print(f"Valid records: {len(s):,}; mean: {s.mean():,.1f} sq ft; median: {s.median():,.1f} sq ft")
    print(f"Min: {s.min():,.1f}; Q1: {s.quantile(.25):,.1f}; Q3: {s.quantile(.75):,.1f}; Max: {s.max():,.1f}")
    sns.histplot(s, bins=30, kde=True); plt.title("Q2. Property size distribution"); plt.xlabel("Size (sq ft)"); plt.show()

# Q3. Price per sq ft by property type
section(3, "How does price per sq ft vary by property type?")
if has("Property_Type", "Price_per_SqFt"):
    t = eda.assign(_psf=numeric("Price_per_SqFt")).groupby("Property_Type")["_psf"].agg(Count="count", Mean="mean", Median="median").sort_values("Median", ascending=False)
    display(t.round(2))
    sns.boxplot(data=eda.assign(_psf=numeric("Price_per_SqFt")), x="Property_Type", y="_psf")
    plt.title("Q3. Stored price-per-sq-ft by property type"); plt.xticks(rotation=35); plt.tight_layout(); plt.show()
    print("Interpretation: compare medians as well as means; confirm the stored field's unit before using monetary labels.")

# Q4. Relationship between size and price
section(4, "Is there a relationship between property size and price?")
if has("Size_in_SqFt", "Price_in_Lakhs"):
    z = eda[["Size_in_SqFt","Price_in_Lakhs"]].apply(pd.to_numeric, errors="coerce").dropna()
    print("Pearson correlation:", round(z.corr(method="pearson").iloc[0,1], 3), "| Spearman correlation:", round(z.corr(method="spearman").iloc[0,1], 3))
    sns.regplot(data=z, x="Size_in_SqFt", y="Price_in_Lakhs", scatter_kws={"alpha":.35})
    plt.title("Q4. Property size vs total price"); plt.show()
    print("Interpretation: positive correlation means larger properties tend to have higher total prices; it does not mean every larger property costs more.")

# Q5. Outliers in price per sq ft or size
section(5, "Are there outliers in price per sq ft or property size?")
for col in ["Price_per_SqFt", "Size_in_SqFt"]:
    if col in eda:
        s = numeric(col).dropna()
        q1, q3 = s.quantile(.25), s.quantile(.75); iqr = q3-q1
        lo, hi = q1-1.5*iqr, q3+1.5*iqr
        print(f"{col}: lower fence={lo:,.2f}, upper fence={hi:,.2f}, flagged={((s<lo)|(s>hi)).sum():,} ({((s<lo)|(s>hi)).mean()*100:.2f}%)")
        print("Extreme records (first 10):")
        display(eda.loc[s.index[(s<lo)|(s>hi)], [c for c in ["City","Locality","Property_Type","Size_in_SqFt","Price_in_Lakhs",col] if c in eda]].head(10))
print("Limit: IQR flags require investigation; do not delete automatically. Genuine luxury/large properties can be valid extremes.")

# Q6. Average price per sq ft by state
section(6, "What is the average price per sq ft by state?")
if has("State", "Price_per_SqFt"):
    display(eda.assign(_psf=numeric("Price_per_SqFt")).groupby("State")["_psf"].agg(Count="count", Mean="mean", Median="median").sort_values("Mean", ascending=False).round(2))
    print("Limit: stored Price_per_SqFt unit must be validated; states with few records have unstable averages.")

# Q7. Average property price by city
section(7, "What is the average property price by city?")
if has("City", "Price_in_Lakhs"):
    city = eda.assign(_price=numeric("Price_in_Lakhs")).groupby("City")["_price"].agg(Count="count", Average="mean", Median="median").sort_values("Average", ascending=False)
    display(city.round(2))
    print("Use median alongside average because expensive outliers can raise the mean.")

# Q8. Median age by locality
section(8, "What is the median age of properties by locality?")
if has("Locality", "Age_of_Property"):
    t = eda.assign(_age=numeric("Age_of_Property")).groupby("Locality")["_age"].agg(Count="count", Median_Age="median").sort_values("Median_Age", ascending=False)
    display(t.round(1))
    print("Limit: locality labels may repeat across cities; if so, group by City and Locality together. Age depends on the reference year used during data preparation.")

# Q9. BHK distribution across cities
section(9, "How is BHK distributed across cities?")
if has("City", "BHK"):
    tab = pd.crosstab(eda["City"], eda["BHK"], margins=False)
    display(tab)
    topcities = eda["City"].value_counts().head(12).index
    display(tab.loc[tab.index.intersection(topcities)])
    print("Limit: counts reflect the dataset sample, not each city's complete housing stock.")

# Q10. Price trends for top 5 most expensive localities
section(10, "What are the price trends for the top 5 most expensive localities?")
if has("Locality", "Price_in_Lakhs"):
    loc = eda.assign(_price=numeric("Price_in_Lakhs")).groupby("Locality")["_price"].agg(Count="count", Mean="mean", Median="median")
    loc = loc[loc["Count"] >= 2].sort_values("Mean", ascending=False).head(5)
    display(loc.round(2))
    print("These are cross-sectional locality comparisons, not time-series trends. A date/year column is needed to establish price movement over time.")
    print("Top 5 by mean price (with at least 2 observations):", list(loc.index))

# Q11. Numeric feature correlation
section(11, "How are numeric features correlated with each other?")
num = eda.select_dtypes(include="number").copy()
if len(num.columns) >= 2:
    corr = num.corr(method="pearson")
    display(corr.round(2))
    plt.figure(figsize=(12,9)); sns.heatmap(corr, cmap="coolwarm", center=0); plt.title("Q11. Numeric feature correlation"); plt.tight_layout(); plt.show()
    print("Limit: correlation is pairwise linear association; IDs, encoded flags, duplicated/derived fields and target-derived columns can create misleading correlations.")
else:
    print("Not enough numeric columns.")

# Q12. Nearby schools vs price per sq ft
section(12, "How do nearby schools relate to price per sq ft?")
if has("Nearby_Schools", "Price_per_SqFt"):
    z = eda[["Nearby_Schools","Price_per_SqFt"]].apply(pd.to_numeric, errors="coerce").dropna()
    print("Pearson r:", round(z.corr().iloc[0,1],3), "| Spearman r:", round(z.corr(method="spearman").iloc[0,1],3))
    display(z.groupby("Nearby_Schools")["Price_per_SqFt"].agg(Count="count", Mean="mean", Median="median").round(2))
    sns.regplot(data=z, x="Nearby_Schools", y="Price_per_SqFt", x_jitter=.08, scatter_kws={"alpha":.4}); plt.title("Q12. Schools vs stored price per sq ft"); plt.show()
    print("Limit: count may not represent school quality, distance, capacity or catchment area; correlation is not causation.")

# Q13. Nearby hospitals vs price per sq ft
section(13, "How do nearby hospitals relate to price per sq ft?")
if has("Nearby_Hospitals", "Price_per_SqFt"):
    z = eda[["Nearby_Hospitals","Price_per_SqFt"]].apply(pd.to_numeric, errors="coerce").dropna()
    print("Pearson r:", round(z.corr().iloc[0,1],3), "| Spearman r:", round(z.corr(method="spearman").iloc[0,1],3))
    display(z.groupby("Nearby_Hospitals")["Price_per_SqFt"].agg(Count="count", Mean="mean", Median="median").round(2))
    sns.regplot(data=z, x="Nearby_Hospitals", y="Price_per_SqFt", x_jitter=.08, scatter_kws={"alpha":.4}); plt.title("Q13. Hospitals vs stored price per sq ft"); plt.show()
    print("Limit: count alone does not capture hospital size, specialty, quality or travel time.")

# Q14. Price by furnished status
section(14, "How does price vary by furnished status?")
if has("Furnished_Status", "Price_in_Lakhs"):
    display(eda.assign(_price=numeric("Price_in_Lakhs")).groupby("Furnished_Status")["_price"].agg(Count="count", Mean="mean", Median="median").sort_values("Median", ascending=False).round(2))
    sns.boxplot(data=eda, x="Furnished_Status", y=numeric("Price_in_Lakhs")); plt.title("Q14. Price by furnished status"); plt.xticks(rotation=25); plt.show()
    print("Limit: differences may reflect city, size, property type and quality rather than furnishing alone.")

# Q15. Price per sq ft by facing direction
section(15, "How does price per sq ft vary by property facing direction?")
if has("Facing", "Price_per_SqFt"):
    display(eda.assign(_psf=numeric("Price_per_SqFt")).groupby("Facing")["_psf"].agg(Count="count", Mean="mean", Median="median").sort_values("Median", ascending=False).round(2))
    sns.boxplot(data=eda, x="Facing", y=numeric("Price_per_SqFt")); plt.title("Q15. Stored price per sq ft by facing"); plt.xticks(rotation=30); plt.show()
    print("Limit: facing categories may be unevenly represented; do not infer a universal directional premium.")

# Q16. Owner type counts
section(16, "How many properties belong to each owner type?")
if has("Owner_Type"):
    counts = eda["Owner_Type"].fillna("Missing").value_counts()
    display(counts.to_frame("Property_Count"))
    print("Percentages:"); display((counts/len(eda)*100).round(2).to_frame("Percent_of_all_rows"))

# Q17. Availability status counts
section(17, "How many properties are available under each availability status?")
if has("Availability_Status"):
    counts = eda["Availability_Status"].fillna("Missing").value_counts()
    display(counts.to_frame("Property_Count"))
    print("Limit: dataset status is only as current as its collection/update date; it may not equal live availability.")

# Q18. Parking space and property price
section(18, "Does parking space affect property price?")
if has("Parking_Space", "Price_in_Lakhs"):
    display(eda.assign(_price=numeric("Price_in_Lakhs")).groupby("Parking_Space")["_price"].agg(Count="count", Mean="mean", Median="median").round(2))
    sns.boxplot(data=eda, x="Parking_Space", y=numeric("Price_in_Lakhs")); plt.title("Q18. Price by parking availability"); plt.show()
    print("Limit: observational comparison only; parking may be associated with larger/newer or higher-end properties.")

# Q19. Amenities and price per sq ft
section(19, "How do amenities affect price per sq ft?")
amenity_cols = [c for c in ["Clubhouse","Garden","Gym","Playground","Pool"] if c in eda.columns]
if not amenity_cols and "Amenities" in eda.columns:
    for a in ["Clubhouse","Garden","Gym","Playground","Pool"]:
        eda["_amenity_"+a] = eda["Amenities"].fillna("").apply(lambda x: int(a.lower() in str(x).lower()))
    amenity_cols = ["_amenity_"+a for a in ["Clubhouse","Garden","Gym","Playground","Pool"]]
if amenity_cols and "Price_per_SqFt" in eda.columns:
    for c in amenity_cols:
        eda[c] = pd.to_numeric(eda[c], errors="coerce").fillna(0).astype(int)
    eda["_amenity_count_eda"] = eda[amenity_cols].sum(axis=1)
    display(eda.groupby("_amenity_count_eda")["Price_per_SqFt"].agg(Count="count", Mean="mean", Median="median").round(2))
    sns.boxplot(data=eda, x="_amenity_count_eda", y=numeric("Price_per_SqFt")); plt.title("Q19. Price per sq ft by amenity count"); plt.xlabel("Number of listed amenities"); plt.show()
    print("Individual amenity comparisons:")
    for c in amenity_cols:
        print("\n", c); display(eda.groupby(c)["Price_per_SqFt"].agg(Count="count", Mean="mean", Median="median").round(2))
    print("Limit: amenity presence is not a measure of quality, maintenance or usage. Amenity-rich properties may also differ by location and size.")
else:
    print("Amenities indicators/text or Price_per_SqFt are unavailable.")

# Q20. Public transport and price / investment potential
section(20, "How does public transport accessibility relate to price per sq ft or investment potential?")
if has("Public_Transport_Accessibility", "Price_per_SqFt"):
    display(eda.assign(_psf=numeric("Price_per_SqFt")).groupby("Public_Transport_Accessibility")["_psf"].agg(Count="count", Mean="mean", Median="median").round(2))
    sns.boxplot(data=eda, x="Public_Transport_Accessibility", y=numeric("Price_per_SqFt")); plt.title("Q20. Transport accessibility vs stored price per sq ft"); plt.xticks(rotation=25); plt.show()
    if "Good_Investment_V2" in eda.columns:
        display(pd.crosstab(eda["Public_Transport_Accessibility"], eda["Good_Investment_V2"], normalize="index").mul(100).round(1).rename(columns={0:"Not good investment %",1:"Good investment %"}))
        print("The investment comparison uses the project's rule-generated target, not realized returns.")
    elif "Investment_Status" in eda.columns:
        display(pd.crosstab(eda["Public_Transport_Accessibility"], eda["Investment_Status"], normalize="index").mul(100).round(1))
    print("Limit: accessibility labels are coarse; actual walking distance, frequency, reliability, commute times and future infrastructure are not measured.")
else:
    print("Required transport or price-per-sq-ft column is unavailable.")

print("\n" + "="*90)
print("EDA COMPLETE — interpret all findings with the limitations above.")
print("="*90)
```
