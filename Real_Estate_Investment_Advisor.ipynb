```python
from sklearn.model_selection import train_test_split

# Columns that must not be used as model inputs
columns_to_remove = [
    "ID",
    "Amenities",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Size_in_SqFt",
    "Good_Investment",
    "Future_Price_5Y"
]

# Create input features and target
X = df_clean.drop(columns=columns_to_remove)
y = df_clean["Good_Investment"]

print("Input features shape:", X.shape)
print("Target shape:", y.shape)

print("\nInput feature columns:")
print(X.columns.tolist())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("Training class distribution:")
display(y_train.value_counts(normalize=True).round(3) * 100)

print("\nTesting class distribution:")
display(y_test.value_counts(normalize=True).round(3) * 100)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Scikit-learn libraries imported successfully!")
```

```python
# Identify categorical columns
categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

# Identify numerical columns
numerical_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

print("Categorical features:", len(categorical_features))
print(categorical_features)

print("\nNumerical features:", len(numerical_features))
print(numerical_features)
```

```python
# Numerical preprocessing
numerical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler(with_mean=False))
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing methods
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
# Create complete Machine Learning pipeline
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=500))
])

print("Logistic Regression model created successfully!")
```

```python
print("Model training started...")

model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
# Predict using testing data
y_pred = model.predict(X_test)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("MODEL PERFORMANCE")
print("-------------------------")
print("Accuracy:", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall:", round(recall * 100, 2), "%")
print("F1 Score:", round(f1 * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=[
        "Not Selected",
        "Potential Investment"
    ],
    cmap="Blues"
)

plt.title("Good Investment - Confusion Matrix")
plt.show()
```

```python
# Check relationship between target and price per sqft

print("Average Price per SqFt by Investment Class:")

print(
    df_clean.groupby("Good_Investment")["Price_per_SqFt"].mean()
)

print("\nMedian Price per SqFt by Investment Class:")

print(
    df_clean.groupby("Good_Investment")["Price_per_SqFt"].median()
)
```

```python
# Calculate city-wise median price per sqft

city_median = df_clean.groupby("City")[
    "Price_per_SqFt"
].transform("median")

# Create improved investment target

df_clean["Good_Investment_V2"] = (
    (df_clean["Price_per_SqFt"] < city_median) &
    (df_clean["Amenities_Count"] >= 2)
).astype(int)

# Display class distribution

print("New Investment Target Distribution:")

print(df_clean["Good_Investment_V2"].value_counts())

print("\nPercentage Distribution:")

print(
    df_clean["Good_Investment_V2"]
    .value_counts(normalize=True).mul(100).round(2)
)
```

```python
# Select input features

columns_to_remove = [
    "ID",
    "Amenities",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Size_in_SqFt",
    "Good_Investment",
    "Good_Investment_V2",
    "Future_Price_5Y"
]

X_v2 = df_clean.drop(columns=columns_to_remove)

# New target variable

y_v2 = df_clean["Good_Investment_V2"]

print("Input features shape:", X_v2.shape)
print("Target shape:", y_v2.shape)

print("\nTarget distribution:")
print(y_v2.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train_v2, X_test_v2, y_train_v2, y_test_v2 = train_test_split(
    X_v2,
    y_v2,
    test_size=0.20,
    random_state=42,
    stratify=y_v2
)

print("Training data:", X_train_v2.shape)
print("Testing data:", X_test_v2.shape)

print("Training completed successfully!")
```

```python
rf_model_v2 = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

print("New Random Forest model created successfully!")
```

```python
print("Training started...")

rf_model_v2.fit(X_train_v2, y_train_v2)

print("Training completed successfully!")
```

```python
rf_pred_v2 = rf_model_v2.predict(X_test_v2)

print("RANDOM FOREST V2 PERFORMANCE")
print("-----------------------------")

print("Accuracy:", round(accuracy_score(y_test_v2, rf_pred_v2) * 100, 2), "%")
print("Precision:", round(precision_score(y_test_v2, rf_pred_v2) * 100, 2), "%")
print("Recall:", round(recall_score(y_test_v2, rf_pred_v2) * 100, 2), "%")
print("F1 Score:", round(f1_score(y_test_v2, rf_pred_v2) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test_v2, rf_pred_v2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Generate confusion matrix
cm_v2 = confusion_matrix(y_test_v2, rf_pred_v2)

# Plot confusion matrix
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_v2,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Good Investment", "Good Investment"],
    yticklabels=["Not Good Investment", "Good Investment"]
)

plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")
plt.title("Random Forest V2 - Confusion Matrix")

plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Get prediction probabilities
rf_prob_v2 = rf_model_v2.predict_proba(X_test_v2)[:, 1]

# Test different probability thresholds
thresholds = [0.50, 0.55, 0.60, 0.65, 0.70, 0.75, 0.80]

print("THRESHOLD COMPARISON")
print("--------------------")

for threshold in thresholds:

    predictions = (rf_prob_v2 >= threshold).astype(int)

    accuracy = accuracy_score(y_test_v2, predictions)
    precision = precision_score(y_test_v2, predictions)
    recall = recall_score(y_test_v2, predictions)
    f1 = f1_score(y_test_v2, predictions)

    print("\nThreshold:", threshold)
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print("Precision:", round(precision * 100, 2), "%")
    print("Recall:", round(recall * 100, 2), "%")
    print("F1 Score:", round(f1 * 100, 2), "%")
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# Calculate ROC-AUC score
auc_score = roc_auc_score(y_test_v2, rf_prob_v2)

print("RANDOM FOREST V2 - ROC AUC SCORE")
print("--------------------------------")
print("ROC-AUC:", round(auc_score, 4))

# Calculate ROC curve
fpr, tpr, thresholds_roc = roc_curve(y_test_v2, rf_prob_v2)

# Plot ROC curve
plt.figure(figsize=(7, 5))

plt.plot(fpr, tpr, label=f"Random Forest V2 (AUC = {auc_score:.4f})")
plt.plot([0, 1], [0, 1], linestyle="--", label="Random Guess")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Random Forest V2")
plt.legend()
plt.grid()

plt.show()
```

```python
# STEP 29: TRAIN CLEAN RANDOM FOREST MODEL

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
from sklearn.base import clone
import joblib

# 1. Identify numerical and categorical columns
numeric_features = X_clean.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_clean.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# 2. Numerical preprocessing
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# 3. Categorical preprocessing
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# 4. Combine preprocessing
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# 5. Create Random Forest pipeline
clean_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=150,
        max_depth=20,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# 6. Split dataset
X_train_clean, X_test_clean, y_train_clean, y_test_clean = train_test_split(
    X_clean,
    y_clean,
    test_size=0.20,
    random_state=42,
    stratify=y_clean
)

print("Training data:", X_train_clean.shape)
print("Testing data:", X_test_clean.shape)

# 7. Train model
print("\nTraining Clean Random Forest Model...")

clean_model.fit(X_train_clean, y_train_clean)

print("Training completed successfully!")

# 8. Predictions
y_pred_clean = clean_model.predict(X_test_clean)

# 9. Evaluation
print("\nMODEL ACCURACY:")
print(accuracy_score(y_test_clean, y_pred_clean) * 100, "%")

print("\nCLASSIFICATION REPORT:")
print(classification_report(y_test_clean, y_pred_clean))

print("\nCONFUSION MATRIX:")
print(confusion_matrix(y_test_clean, y_pred_clean))

# 10. Save model
joblib.dump(clean_model, "real_estate_clean_model.pkl")

print("\nClean model saved successfully!")
```

```python
# STEP 30: REMOVE INDIRECT TARGET LEAKAGE

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
import joblib

leakage_features = [
    "Price_in_Lakhs",
    "Size_in_SqFt",
    "Amenities",
    "Clubhouse",
    "Garden",
    "Gym",
    "Playground",
    "Pool"
]

# Create stricter feature dataset
X_strict = X_clean.drop(
    columns=leakage_features,
    errors="ignore"
)

print("STRICT DATASET SHAPE:", X_strict.shape)
print("\nREMAINING FEATURES:")
print(X_strict.columns.tolist())

# Split dataset
X_train_strict, X_test_strict, y_train_strict, y_test_strict = train_test_split(
    X_strict,
    y_clean,
    test_size=0.20,
    random_state=42,
    stratify=y_clean
)

# Re-extract numeric and categorical feature lists for strict model
numeric_features_strict = X_strict.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features_strict = X_strict.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Rebuild preprocessing blocks to exclude leakage columns
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_strict = ColumnTransformer([
    ("num", numeric_transformer, numeric_features_strict),
    ("cat", categorical_transformer, categorical_features_strict)
])

# Create fresh Strict Random Forest Pipeline
strict_model = Pipeline([
    ("preprocessor", preprocessor_strict),
    ("classifier", RandomForestClassifier(
        n_estimators=150,
        max_depth=20,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# Train
print("\nTraining Strict Random Forest Model...")
strict_model.fit(X_train_strict, y_train_strict)

# Predict
y_pred_strict = strict_model.predict(X_test_strict)

# Evaluate
print("\nSTRICT MODEL ACCURACY:")
print(accuracy_score(y_test_strict, y_pred_strict) * 100, "%")

print("\nCLASSIFICATION REPORT:")
print(classification_report(y_test_strict, y_pred_strict))

print("\nCONFUSION MATRIX:")
print(confusion_matrix(y_test_strict, y_pred_strict))

# Save separately
joblib.dump(strict_model, "real_estate_strict_model.pkl")
print("\nStrict model saved successfully!")
```

```python
# Step 1: Calculate price score
advisor_df["Price_Advantage_Score"] = (
    advisor_df["Price_per_SqFt"] < city_median
).astype(int) * 60

# Step 2: Calculate amenities score
advisor_df["Amenities_Score"] = (
    advisor_df["Amenities_Count"].clip(0, 5) * 8
)

# Step 3: Calculate total investment score
advisor_df["Investment_Score"] = (
    advisor_df["Price_Advantage_Score"] +
    advisor_df["Amenities_Score"]
)

# Step 4: Assign investment category
advisor_df["Investment_Category"] = advisor_df[
    "Investment_Score"
].apply(
    lambda score:
        "High Potential" if score >= 76
        else "Moderate Potential" if score >= 60
        else "Low Potential"
)

# Step 5: Display top 10 properties
top_properties = advisor_df.sort_values(
    by="Investment_Score",
    ascending=False
)

print("PROPERTY INVESTMENT SCORING SYSTEM")
print("-----------------------------------")

print("\nAverage Investment Score:",
      round(advisor_df["Investment_Score"].mean(), 2))

print("\nInvestment Category Summary:")
print(advisor_df["Investment_Category"].value_counts())

print("\nTOP 10 INVESTMENT PROPERTIES:")

display(
    top_properties[
        ["City", "Locality", "Property_Type",
         "Price_in_Lakhs", "Amenities_Count",
         "Investment_Score", "Investment_Category"]
    ].head(10)
)
```
